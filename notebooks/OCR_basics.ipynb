# OCR Fundamentals & PaddleOCR

## 1. What is OCR?

**OCR (Optical Character Recognition)** is a technology used to extract text from images, scanned documents, and other visual sources and convert it into machine-readable text.

In the VIGILOX Document Intelligence project, OCR is responsible for converting security licenses, ID cards, and SIA badges into raw text that can later be processed by an LLM.

```text
Document Image
      ↓
     OCR
      ↓
  Extracted Text
```

---

## 2. How Does OCR Work?

A typical OCR pipeline consists of several stages:

```text
Input Image
     ↓
Image Preprocessing
     ↓
Text Detection
     ↓
Text Recognition
     ↓
Text + Confidence + Bounding Boxes
```

### Text Detection

Identifies **where text is located** in the image.

### Text Recognition

Identifies **what the detected text actually says**.

For example:

```text
Image
  ↓
Text Detection
  ↓
Text Region
  ↓
Text Recognition
  ↓
"John Smith"
```

---

## 3. Tesseract vs PaddleOCR

Both **Tesseract** and **PaddleOCR** are OCR technologies used to extract text from images and documents.

| Feature              | Tesseract          | PaddleOCR          |


# OCR Fundamentals

## 1. What is OCR?

**OCR = Optical Character Recognition**

OCR is a technology that extracts text from an image or scanned document and converts it into **machine-readable text**.

### Simple Example

```text
📄 Image
   ↓
  OCR
   ↓
"Name: John Smith
 License: ABC12345
 Expiry: 15/08/2027"
```

Normally, a computer sees an image as a collection of pixels:

```text
🖼️ → Pixels
```

OCR identifies letters and words within those pixels and converts them into text:

```text
🖼️ → OCR → "John Smith"
```

### In the VIGILOX Project

When a security guard license is uploaded:

```text
License Image
      ↓
     OCR
      ↓
   Raw Text
      ↓
     LLM
      ↓
Structured Data
```

**The primary job of OCR is to extract text from the document.**

---

# 2. How Does OCR Work?

Modern OCR can be understood as a pipeline consisting of several stages:

```text
Image
  ↓
Preprocessing
  ↓
Text Detection
  ↓
Text Recognition
  ↓
Text + Confidence + Bounding Boxes
```

### Step 1 — Input Image

The system receives a document image:

```text
┌──────────────────────┐
│ SECURITY LICENSE     │
│                      │
│ Name: John Smith     │
│ License: ABC12345    │
│ Expiry: 15/08/2027   │
└──────────────────────┘
```

### Step 2 — Image Preprocessing

The image can be improved before OCR processing:

```text
Resize
Grayscale
Denoise
Threshold
Deskew
```

The purpose is to make the document easier for the OCR model to process.

### Step 3 — Text Detection

The OCR system identifies **which areas of the image contain text**.

For example:

```text
┌──────────────────────────┐
│ [SECURITY LICENSE]       │
│                          │
│ [Name: John Smith]       │
│ [License: ABC12345]      │
│ [Expiry: 15/08/2027]     │
└──────────────────────────┘
```

### Step 4 — Text Recognition

The detected text regions are then recognized:

```text
Name: John Smith
License: ABC12345
Expiry: 15/08/2027
```

### Step 5 — OCR Output

An OCR system can provide more than just text:

```text
Text
Confidence
Bounding Box
```

---

# 3. Text Detection vs. Text Recognition

This is an **important distinction** in OCR.

## Text Detection

Text detection answers:

> **"Where is the text located in the image?"**

Example:

```text
Image
┌─────────────────────────┐
│                         │
│  ┌───────────────────┐  │
│  │ John Smith        │  │
│  └───────────────────┘  │
│                         │
└─────────────────────────┘
```

The detection system identifies the location of the text.

For example:

```text
x1, y1, x2, y2
```

These coordinates define the detected text region.

---

## Text Recognition

Text recognition answers:

> **"What does the detected text actually say?"**

Example:

```text
Detected Region
      ↓
Text Recognition
      ↓
"John Smith"
```

### Together

```text
Image
  ↓
Text Detection
  ↓
"Text is located here"
  ↓
Text Recognition
  ↓
"John Smith"
```

### Easy Example

Think of a person looking at a document.

**Detection:**

> "There are 5 lines of text on this page."

**Recognition:**

> "The first line says John Smith."

---

# 4. What is OCR Confidence?

OCR does not always recognize text with 100% certainty.

Therefore, OCR systems can provide a **confidence score** indicating how confident the model is in its recognition result.

For example:

```text
John Smith     → 0.98
ABC12345       → 0.94
15/08/2027     → 0.87
```

Generally:

```text
0.98 → Very high confidence
0.94 → High confidence
0.87 → Lower confidence
```

The exact confidence scale depends on the OCR library. It may be represented as:

```text
0–1
```

or:

```text
0–100%
```

### Why Can Confidence Be Low?

Common reasons include:

* Blurry images
* Very small text
* Poor lighting
* Rotated documents
* Noisy backgrounds
* Unusual fonts
* Handwritten text
* Low-resolution images

For example, the OCR system may incorrectly interpret:

```text
ABC12345
```

as:

```text
ABC12?45
```

### Why Is This Important for VIGILOX?

Suppose the OCR extracts:

```text
License Number: ABC12345
Confidence: 0.62
```

The system can flag it for human review:

```text
Low Confidence
      ↓
Human Review
```

This becomes an important foundation for the project's **human review queue**.

# 5. What Are Bounding Boxes?

A **bounding box** is a rectangular area around a detected text region or object in an image.

Example:

```text
┌─────────────────────────────┐
│                             │
│   ┌─────────────────────┐   │
│   │ John Smith          │   │
│   └─────────────────────┘   │
│                             │
└─────────────────────────────┘
```

The rectangle surrounding `John Smith` represents its **bounding box**.

Bounding boxes are usually represented using coordinates.

```text
(x1, y1)
      ┌──────────────┐
      │ John Smith   │
      └──────────────┘
                 (x2, y2)
```

An OCR system may return something conceptually similar to:

```text
Text: "John Smith"

Bounding Box:
[
    [120, 80],
    [300, 80],
    [300, 120],
    [120, 120]
]
```

> The exact bounding-box format depends on the OCR library.

### Why Are Bounding Boxes Useful?

In the VIGILOX project, bounding boxes can be used to:

* Highlight extracted text on the original document
* Show reviewers where information was detected
* Visually inspect OCR results
* Identify specific document regions
* Support document field localization

# 6. How Does Image Preprocessing Improve OCR?

Real-world documents are not always perfect.

An image might be:

* Blurry
* Dark
* Rotated
* Noisy
* Low-resolution
* Poorly scanned

Without preprocessing:

```text
Original Image
      ↓
Blurry / Dark / Noisy
      ↓
OCR
      ↓
Poor Extraction
```

The goal of preprocessing is to make the image **easier for the OCR system to read**.

---

## A. Grayscale

Grayscale converts a color image into shades of gray.

```text
RGB Image
    ↓
Grayscale Image
```

This can remove unnecessary color information and simplify the image.

---

## B. Resizing

If text is very small:

```text
Small Text
    ↓
Resize / Enlarge
    ↓
Larger Text
```

Larger and clearer characters can make recognition easier.

---

## C. Thresholding

Thresholding converts an image into a simplified representation, often approximately:

```text
Black Text
+
White Background
```

Example:

```text
Gray Image
    ↓
Thresholding
    ↓
Black & White Image
```

This can be particularly useful for clean scanned documents.

---

## D. Noise Removal

Images may contain unwanted visual artifacts:

```text
.... John Smith .....
```

Noise removal attempts to reduce these unwanted artifacts so that the text becomes clearer.

---

## E. Deskewing

If a document is slightly tilted:

```text
//////// Text ////////
```

deskewing rotates the image to straighten the text:

```text
John Smith
License: ABC12345
```

This can make text recognition easier.
